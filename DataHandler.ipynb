```python
from pathlib import Path
import pandas as pd

corpus_dir = Path("corpus")
metadata = pd.read_csv(corpus_dir / "metadata.csv")
metadata.groupby(["label", "source_name"])["n_words"].agg(["count", "mean", "std", "min", "max", "sum"])
```

```python
assert len(metadata) == 300
assert metadata["label"].value_counts().to_dict() == {0: 150, 1: 150}
assert all((corpus_dir / ("Falso" if row.label == 0 else "Verdad") / f"{row.doc_id}.txt").exists() for row in metadata.itertuples())
print("Corpus válido: 300 documentos, 150 por clase.")
```
